## 📊 Step 6: Render Interactive Visualizations (Stage 3 Grammar of Graphics Specs)

```python
import plotly.express as px

charts = results['stage_3_charts']
print(f'Generated {len(charts)} Interactive Plotly Charts from Stage 3 JSON specs:')
for i, c in enumerate(charts, 1):
    print(f"{i}. {c['title']} ({c['chart_type'].upper()})")

# Render Chart 1: Bar Chart
cat_summary = df.groupby('category')['units_sold'].sum().reset_index()
fig1 = px.bar(cat_summary, x='category', y='units_sold', title='Total Transaction Volume by Category', color='category')
fig1.show()

# Render Chart 2: Scatter Plot with $18k Outlier
fig2 = px.scatter(df, x='units_sold', y='revenue', title='Revenue vs Units Sold (Volume Sensitivity)', hover_data=['order_id'])
fig2.show()
```
